```python
# ==========================================
# Job Scheduling using Graph (Adjacency List & Traversal)
# Syllabus Alignment: Storage Representation (Adjacency List) & Graph Traversal
# ==========================================

def job_scheduling(vertices, edges):
    # 1. Storage Representation: Adjacency List and In-Degree tracking
    adj_list = {v: [] for v in vertices}
    in_degree = {v: 0 for v in vertices}

    # Populate the Adjacency List and calculate in-degrees (incoming edges)
    for u, v in edges:
        adj_list[u].append(v)
        in_degree[v] += 1

    # 2. Graph Traversal Initialization (Queue for nodes with 0 incoming edges)
    queue = []
    for v in vertices:
        if in_degree[v] == 0:
            queue.append(v)

    execution_order = []

    # 3. Graph Traversal Process (BFS-style processing)
    while len(queue) > 0:
        current_vertex = queue.pop(0)  # Dequeue the current job
        execution_order.append(current_vertex)

        # Traverse all adjacent neighbors (outgoing edges)
        for neighbor in adj_list[current_vertex]:
            in_degree[neighbor] -= 1  # Reduce incoming edge count
            
            # If all prerequisites are met (in-degree becomes 0), add to traversal queue
            if in_degree[neighbor] == 0:
                queue.append(neighbor)

    # 4. Cycle Detection Check (Deadlock verification)
    if len(execution_order) == len(vertices):
        return execution_order
    else:
        return "Error: Cycle detected in graph! Scheduling failed due to a deadlock."


# ==========================================
# Test Execution
# ==========================================
if __name__ == "__main__":
    # Graph Vertices (Jobs/Tasks)
    jobs = ["Task A", "Task B", "Task C", "Task D", "Task E"]

    # Directed Edges representing prerequisites: (Prerequisite -> Dependent Job)
    dependencies = [
        ("Task A", "Task C"),  # Task A must finish before Task C
        ("Task B", "Task C"),  # Task B must finish before Task C
        ("Task C", "Task D"),  # Task C must finish before Task D
        ("Task D", "Task E")   # Task D must finish before Task E
    ]

    print("--- Graph-Based Job Scheduling Sequence ---")
    result = job_scheduling(jobs, dependencies)

    if isinstance(result, list):
        for step, task in enumerate(result, 1):
            print(f"Step {step}: {task}")
    else:
        print(result)
```
